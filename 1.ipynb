# Problem Statement: Write a Python program to find the maximum value from a list of numbers using a user-defined function.
# Aim: To implement a Python function that returns the maximum element from a list while handling edge cases.

```python
def find_max_value(numbers):
    if len(numbers) == 0:
        return None
    return max(numbers)

numbers = list(map(int, input("Enter numbers: ").split()))

result = find_max_value(numbers)

if result is None:
    print("List is empty")
else:
    print("Maximum value is:", result)
```
